# Sea Level Projection - Datasheet Analysis

Charting sea level projections from 2005 to 2150. Data: NOAA 2022 Sea Level Rise Technical Report.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt

DATA = os.path.join(os.environ.get('DATA_DIR', '/data'), 'slp_data.csv')
df = pd.read_csv(DATA)
df[['scenario', 'level']] = df['Scenario'].str.split(' - ', expand=True)

year_cols = [c for c in df.columns if c.startswith('RSL') and c[3:7].isdigit()]
years = [int(c[3:7]) for c in year_cols]

# Named stations only (the rest are interpolated grid cells)
stations = df[df['NOAA Name'].notna()]
stations[['NOAA Name', 'scenario', 'level'] + year_cols].head()
```

```python
def plot_station(name, level='MED'):
    sub = stations[(stations['NOAA Name'] == name) & (stations['level'] == level)]
    fig, ax = plt.subplots(figsize=(9, 5))
    for _, row in sub.iterrows():
        ax.plot(years, row[year_cols].astype(float), label=f"{row['scenario']} m")
    ax.set(title=f'{name}: projected relative sea level rise ({level})',
           xlabel='Year', ylabel='Rise since 2005 (cm)')
    ax.legend(title='Scenario (global rise by 2100)')
    plt.show()

plot_station('Seattle')
```
